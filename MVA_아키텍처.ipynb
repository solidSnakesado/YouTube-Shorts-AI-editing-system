- MVA(Minimum Viable Architecture, 최소 실행 가능 아키텍처)는 소프트웨어 개발 초기 단계에서 얼마만큼의 설계가 지금 필요한가를 판단하는 의사결정 프레임워크, 특정 기술 스택이나 디자인 패턴이 아니라, 아키텍처에 대한 사고방식에 가까움

- 등장 배경
    - 소프트웨어 프로젝트가 실패하는 아키텍처의 원인은 보통 두 가지 극단으로 비롯
    - 첫 번째는 설계의 부재, 시간 압박을 이유호 구조 없이 코드를 쌓아 올리면, 프로토타입이 시장에서 성공하고 그 위에 기능이 계속 추가되면서, 시스템은 점점 수정 불가능한 상태로 경직됨, 이것을 "희생적 아키텍처"라 부르는데, "임시적인 해결책만큰 영구적인 것은 없다"는 소프트웨어 업계의 격언이 정확히 이 상화을 묘사
    - 두 번쨔는 과잉 설계, 아직 사용자가 한 명도 없는 시점에 마이크로서비스, 이벤트 소싱, CQRS 같은 대규모 분산 시스템용 패턴을 도입, 설계와 인프라 구축에 대부분의 시간을 소비하고, 정작 시장 검증을 위한 작동하는 제품은 나오지 않음

    - MVA는 이 두 극단 사이에서 적잘한 수준의 설계를 찾는 개면, 현재의 요구사항을 충족하면서도, 미래의 변화에 유연하게 대응할 수 있는 최소한의 아키텍처 결정만을 내리는 것을 목표로 함

- 핵심 정의
    - MVA를 구성하는 구 가지 핵심 단어는 "최소한"과 "설계 결정"
    - "최소한"이란 지금 당장 필요하지 않은 구조를 만들지 않겠다는 뜻, 사용자가 100명도 안되는 제품에 로드 밸런서를 도입하거나, 데이터가 단일 테이블로 충분한데 분산 데이터베이스를 설계하는 것은 MVA의 정신에 반하는 것
    - "설계 결정"이란 코드의 양이 아니라 코드의 위치와 관게에 대한 원칙을 세우는 것, 어떤 로직이 어느 계층에 속하는지 모듈 간 의존성이 어떤 방향으로 흐르는 지, 외부 시스템과의 경계를 어디에 두는 지를 명확이 정의, 코드를 많이 쓰는 것이 아니라 "이 코드는 어디에 있어야 하는 가"에 대한 규칙을 합의 하는 것이 MVA의 본질

- 판단 기준: 품질 속성 식별
    - MVA를 구축할 때 가장 먼저 수행하는 작업은 프로젝트에서 나중에 추가하기 어려운 속성이 무엇인지 식별하는 것
    - 기능(Feature)은 나중에 추가하기 비교적 쉬움, "좋아요 버튼"이 없으면 나중에 만들면 됨, 하지만 보안, 확장성, 성능, 관측가능성(Observability) 같은 비기능적 품질 속성은 시스템이 이미 구축된 후에 끼워 넣기가 매우 어려움, 예를 들어 인증 체계없이 수십개의 엔드포인트를 만든 뒤 나중에 보안을 적용하려면, 모든 엔드 포인트를 하나씩 수정해야 함, 동기식으로 설계된 시스템을 나중에 비동기로 전환하려면 사실상 전체를 다시 작성해야 함
    - 따라서 MVA는 이런 질문을 던짐, "이 프로젝트에서 나중에 변경하면 비용이 극도로 높아지는 결정은 무엇인가?" 그 결정만 초기에 내리고, 나머지는 더 많은 정보가 확보될 때까지 미룸, MVA의 핵심 원리

- MVP와의 관계
    - MVP(Minimum Viable Producr)와 MVA는 서로 다른 층위의 개념
    - MVP는 "사용자에서 보여줄 최소 기능"을 정의, 시장의 반응을 검증하기 위해 핵심 가치만 담은 제품을 가장 빠르게 출시하는 것이 목표
    - MVA는 "그 기능을 담을 최소 구조"를 정의 MVP가 무엇을 만들지에 대한 답이라면, MVA는 어떻게 만들지에 대한 답
    - 둘은 반드시 함께 가야 함, MVP없는 MVA는 사용자 없는 빈 그릇, MVA없는 MVP는 기초없이 세운 건물, MVP가 시장에서 성공했을 때, MVA가 있으면 그 결과물을 폐기하지 않고 실제 제품으로 진화시킬 수 있음, MVA가 없으면 성공 그 자체가 독이 되며, 급증하는 사용자와 기능 요구를 감당하지 못하고 시스템이 무너짐

- 기술 부채에 대한 입장
    - MVA는 기술 부채를 제거하는 도구가 아님, 기술 부채를 두 종류로 분류하고, 전략적으로 관리하는 도구
    - 첫 번째 종류는 아키텍처적 기술 부채(Architectural Technical Debt), 잘못된 프레임워크 선택, 계층 간 경계 없는 결합, 보안 구조의 부재, 관심사가 분리되지 않은 단일 모듈 등이 여기에 해당, 이 부채는 "이자율"이 매우 높음, 시간이 지날수록 수정 비용이 기하급수적으로 증가, 종종 시스템 전체를 처음부터 다시 작성하는 것 외에는 해결 방법이 없음, MVA는 바로 이 부채가 발생하지 않도록 초기에 구조적 원칙을 세움
    - 두 번째 종류는 코드 수준의 기술 부채, 하드코딩된 상수, 불완전한 에러 처리, 부족한 테스트 커버리지, 최적화되지 않은 쿼리 등이 여기에 해당, 이 부채는 해당 파일이나 함수만 수정하면 되므로 이자율이 낮음, MVA는 이런 부터를 프로토타입 단계에서 의도적으로 허용, 즉, MVA의 전략은 "비싼 부채는 지금 방지하고, 싼 부채는 나중에 갚겠다"로 요약, 어떤 부채가 존재하는지, 그것을 언제 갚을 것인지를 팀이 인지하고 문서화하는 것이 무분별하게 부채를 축적하는 것과는 결정적 차이

- 희생적 아키텍처와의 차이
    - 희생적 아키텍처는 "나중에 전부 버리고 다시 만들 것"을 전재함, 현재의 속도를 극대화하기 위해 구조를 의도적으로 무시
    - MVA는 "나중에 확장할 수 있는 토대를 지금 최소한으로 마련하는 것"을 전제, 현재의 속도와 미래의 지속 가능성 사이에서 균형을 잡음

        | 구분 | 희생적 아키텍처 | MVA | 
        | --- | --- | --- |
        | 전제 | 나중에 폐기하고 재작성 | 진화적으로 확장 |
        | 설계 투자 | 거의 없음 | 핵심 품질 속성에 집중한 최소 설계 |
        | 기술 부채 관리 | 무분별 축적 | 분류 후 전략적 허용 | 
        | 성공 시 시나리오 | 성공이 독이 됨 (확장 불가) | 프로토타입이 제품으로 진화 |
        | 적합한 상황 | 순수한 개념 검증 데모 | 제품으로 발전시킬 가능성이 있는 프로토타입 |

    - 희생적 아키텍처가 적합한 경우도 분명 있음, 단순히 UI흐름을 보여주는 시각적 데모나, 기술과 무관한 시장 반응을 테스트하는 경우에도 노코드 도구나 구조 없는 스크립트가 더 효율적 하지만 기술적 타당성을 검증하고 행후 실제 제품으로 반전시켜야 하는 상황이라면, MVA가 필수적

- MVA가 결정하는 것과 미루는 것
    - MVA의 실천적 핵심은 지금 결정할 것과 나중에 결정할 것을 명확히 구분하는 데 있음
    - 지금 결정하는 것(변경 비용이 높은 결정): 시스템의 계층 구조와 모듈 간 경계, 의존성의 방향과 주입 방식, 데이터 접근의 추상화 수준, 인증/인가 메커니즘이 위치할 자리, 동기/비동기 처리 모델 선택, 핵심 데이터 모델의 관계 구조
    - 나중에 결정하는 것(변경 비용이 낮거나, 정보가 더 필요한 결정): 구처젝인 데이터베이스 제품 선택, UI 프레임워크나 디자인 시스템, 캐싱 전략의 세부 사항, 배포 인프라와 CI/CD 파이프라인, 모니터링 도구의 종류, 구체적인 알고리즘이나 라이브러리
    - 이 구분의 기준은 항상 "이 결정을 나중에 바꾸려면 얼마나 많은 코드를 수정해야 하는 가"임, 수정 범위가 한두 파일이면 미루고, 스스템 전반에 걸쳐 있으면 지금 결정

- 핵심 원칙 요약 
    - MVA를 관동하는 사고방식은 결국 세가지로 압축
        - 첫째: 아키텍처는 속도를 늦추는 장애물이 아니라 지속 가능한 속도를 유지하게 해주는 엔진, 초기에 구조를 잡는 시간이 길어 보이지만, 이후 버그 수정 시간 단축, 협업 효율 향상, 기능 추가의 예측 가능성으라는 형태로 돌아옴
        - 둘쩨: 모든 것을 지금 설계하지 않되, 나중에 바꾸기 비싼 것만 지금 설계, 이것이 "최소한"의 의미
        - 셋째: 의도적인 부채와 무분별한 부채는 완전히 다름, 어떤 타협을 했는지 팀이 알고, 언제 갚을 지 계획이 있다면 그것은 관리되는 부채, 아무도 모르는 사이에 쌓인 부채만이 시스템을 무너뜨림
